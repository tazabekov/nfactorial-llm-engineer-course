# Извлечение текста с помощью pypdf

**Задача:** Извлечь текст из финансового отчета Казтелеком, используя простую библиотеку `pypdf`.

*   **Библиотека:** `pypdf`
*   **Документ:** `raw/kaztelecom.pdf`
*   **Страницы:** 11, 12 (индексы 10, 11)
*   **Формат ответа:** JSON
*   **Экспорт:** `01-pypdf-output.json`

```python
%pip install pypdf
```

```python
import os
import json
from pypdf import PdfReader

# Определяем путь к файлу напрямую в папке raw
pdf_path = "raw/kaztelecom.pdf"

# Инициализация PdfReader
reader = PdfReader(pdf_path)

# Извлечение страниц 11 и 12 (0-indexed: 10 и 11)
pages_to_extract = [11, 12]
extracted_pages = []

for page_num in pages_to_extract:
    # Индекс в pypdf на 1 меньше номера страницы
    page_idx = page_num - 1
    page = reader.pages[page_idx]
    text = page.extract_text()
    
    extracted_pages.append({
        "page": page_num,
        "text": text
    })
    print(f"Успешно извлечена страница {page_num} (длина текста: {len(text)} символов)")

# Формирование итогового JSON
output_data = {
    "filename": "kaztelecom.pdf",
    "pages": extracted_pages
}

# Экспорт в файл
output_filename = "01-pypdf-output.json"
with open(output_filename, "w", encoding="utf-8") as f:
    json.dump(output_data, f, ensure_ascii=False, indent=2)

print(f"\nДанные успешно экспортированы в {output_filename}")
```

```python
# Проверка созданного JSON файла
with open("01-pypdf-output.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)

print(f"Имя файла в JSON: {loaded_data['filename']}")
for page_data in loaded_data['pages']:
    print(f"Страница {page_data['page']}: {page_data['text'][:150].strip()}...")
```
