# Демонстрация ограничений pypdf при извлечении текста

Сравним как pypdf справляется с простым текстом vs сложными таблицами

```python
!pip install pypdf
```

```python
# Загрузка файлов в Google Colab
from google.colab import files

print("Загрузите оба PDF файла:")
print("1. NIRM03_SelfLearning_ASR_Kazakh_Updated.pdf")
print("2. 09_2025_Consolidated Financial statements_IFRS_RUS.pdf")
uploaded = files.upload()
```

```python
# Пример 1: Простой текстовый документ (страница 3)
# Здесь pypdf справляется хорошо

from pypdf import PdfReader

reader = PdfReader("NIRM03_SelfLearning_ASR_Kazakh_Updated.pdf")
page = reader.pages[2]  # индексация с 0, поэтому страница 3 = индекс 2

text = page.extract_text()
print("=" * 50)
print("ПРОСТОЙ ТЕКСТ - страница 3")
print("=" * 50)
print(text)
```

```python
# Пример 2: Сложная таблица из финансовой отчетности (страница 23)
# Здесь pypdf теряет структуру - столбцы смешиваются, данные путаются

from pypdf import PdfReader

reader = PdfReader("09_2025_Consolidated Financial statements_IFRS_RUS.pdf")
page = reader.pages[22]  # индексация с 0, поэтому страница 23 = индекс 22

text = page.extract_text()
print("=" * 50)
print("СЛОЖНАЯ ТАБЛИЦА - страница 23")
print("=" * 50)
print(text)
print("\n")
print("=" * 50)
print("ВЫВОД: Структура таблицы полностью потеряна!")
print("Столбцы перемешались, невозможно понять какие данные к чему относятся.")
print("Для сложных документов нужны специализированные инструменты.")
print("=" * 50)
```

# Docling - специализированный инструмент для сложных документов

Теперь посмотрим как Docling справляется с теми же документами

```python
!pip install docling
```

```python
# Docling: Простой текстовый документ (страница 3)
from docling.document_converter import DocumentConverter

converter = DocumentConverter()
result = converter.convert("NIRM03_SelfLearning_ASR_Kazakh_Updated.pdf")

# Получаем markdown весь документ, но можно работать с отдельными элементами
markdown_text = result.document.export_to_markdown()

print("=" * 50)
print("DOCLING - ПРОСТОЙ ТЕКСТ")
print("=" * 50)
print(markdown_text[:3000])  # первые 3000 символов для демонстрации
```

```python
# Docling: Сложная таблица из финансовой отчетности
# Docling сохраняет структуру таблиц!

from docling.document_converter import DocumentConverter

converter = DocumentConverter()
result = converter.convert("09_2025_Consolidated Financial statements_IFRS_RUS (1)-pages.pdf")

markdown_text = result.document.export_to_markdown()

print("=" * 50)
print("DOCLING - СЛОЖНАЯ ТАБЛИЦА")
print("=" * 50)
print(markdown_text)
print("\n")
print("=" * 50)
print("ВЫВОД: Docling сохраняет структуру таблиц!")
print("Таблицы экспортируются в markdown формате с правильными столбцами.")
print("=" * 50)
```
