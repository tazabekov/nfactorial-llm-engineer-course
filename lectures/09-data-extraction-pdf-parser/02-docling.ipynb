# Извлечение текста с помощью Docling (IBM)

**Задача:** Извлечь данные из финансового отчета КазахТелекома, используя инструмент нового поколения — Docling.

*   **Библиотека:** `docling` (IBM)
*   **Документ:** `raw/kaztelecom.pdf`
*   **Страницы:** 11, 12, 16
*   **Цель:** Посмотреть, как Docling справится со структурой документа (таблицами, заголовками) по сравнению с простым pypdf.
*   **Формат ответа:** JSON
*   **Экспорт:** `02-docling-output.json`

```python
# Установка библиотеки docling (если не установлена)
%pip install docling
```

```python
import os
import json
from docling.document_converter import DocumentConverter

# Определяем путь к файлу напрямую в папке raw
pdf_path = "raw/kaztelecom.pdf"

print(f"Используемый путь к файлу: {pdf_path}")

# Инициализация DocumentConverter
converter = DocumentConverter()

# Конвертация документа
result = converter.convert(pdf_path)
doc = result.document

# Страницы для извлечения (1-indexed в docling)
pages_to_extract = [11, 12, 16]
extracted_pages = []

for page_num in pages_to_extract:
    # Извлекаем markdown представление конкретной страницы
    text = doc.export_to_markdown(page_no=page_num)
    
    extracted_pages.append({
        "page": page_num,
        "text": text
    })
    print(f"Успешно извлечена страница {page_num} с помощью Docling (длина текста: {len(text)} символов)")

# Формирование итогового JSON
output_data = {
    "filename": "kaztelecom.pdf",
    "pages": extracted_pages
}

# Экспорт в файл
output_filename = "02-docling-output.json"
with open(output_filename, "w", encoding="utf-8") as f:
    json.dump(output_data, f, ensure_ascii=False, indent=2)

print(f"\nДанные успешно экспортированы в {output_filename}")
```

```python
# Проверка созданного JSON файла
with open("02-docling-output.json", "r", encoding="utf-8") as f:
    loaded_data = json.load(f)

print(f"Имя файла в JSON: {loaded_data['filename']}")
for page_data in loaded_data['pages']:
    print(f"\n--- Страница {page_data['page']} (первые 200 символов) ---")
    print(page_data['text'][:200].strip() + "...")
```
